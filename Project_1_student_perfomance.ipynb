# Data Vizualization

```python
# Count students with a final grade of zero
n_zero = (df["G3"] == 0).sum()

print(f"Students with G3 = 0: {n_zero} ({n_zero / len(df):.1%})")
navy = "#202A44"          # Pinterest-style Navy
atlantic_navy = "#13336F" # Atlantic Navy
complementary = "#6C8EBF" # Soft blue
background = "#F7F8FC"    # Light background
mean_grade = df["G3"].mean()
median_grade = df["G3"].median()
fig = px.histogram(
    df,
    x="G3",
    nbins=21,
    title="Distribution of Final Grades (G3)",
    labels={
        "G3": "Final Grade (0–20)",
        "count": "Number of Students"
    },
    color_discrete_sequence=[atlantic_navy],
    template="plotly_white",
    hover_data=["G3"]
)
fig.add_vline(
    x=mean_grade,
    line_dash="dash",
    line_color="#D94F70",
    line_width=2,
    annotation_text=f"Mean: {mean_grade:.1f}",
    annotation_position="top"
)
fig.add_vline(
    x=median_grade,
    line_dash="dash",
    line_color=complementary,
    line_width=2,
    annotation_text=f"Median: {median_grade:.1f}",
    annotation_position="top left"
)
fig.update_layout(
    title={
        "text": "Distribution of Final Grades (G3)",
        "x": 0.5,
        "xanchor": "center",
        "font": {"size": 22, "color": navy}
    },
    xaxis_title="Final Grade (0–20)",
    yaxis_title="Number of Students",
    plot_bgcolor=background,
    paper_bgcolor=background,
    font=dict(
        family="Arial",
        size=13,
        color=navy
    ),
    bargap=0.08,
    legend_title_text="Statistics",
    margin=dict(l=60, r=40, t=90, b=60)
)
fig.update_xaxes(
    range=[0, 20],
    dtick=1,
    showgrid=True,
    gridcolor="#E1E5EF"
)
fig.update_yaxes( showgrid=True, gridcolor="#E1E5EF")
fig.show()
```

```python
# Calculate the correlation between G2 and G3
corr = df["G2"].corr(df["G3"])
print(f"Correlation between G2 and G3: {corr:.2f}")
navy = "#202A44"
atlantic_navy = "#13336F"
complementary_blue = "#6C8EBF"
accent_pink = "#D94F70"
background = "#F7F8FC"
fig = px.scatter(
    df,
    x="G2",
    y="G3",
    title=f"G2 vs. G3 (Correlation = {corr:.2f})",
    labels={
        "G2": "G2 (Second Period Grade)",
        "G3": "G3 (Final Grade)"
    },
    trendline="ols",
    opacity=0.5,
    color_discrete_sequence=[atlantic_navy],
    template="plotly_white",
    hover_data=["G2", "G3"]
)
fig.update_traces(
    selector=dict(mode="lines"),
    line=dict(color=accent_pink, width=3)
)
fig.update_layout(
    title={
        "text": f"G2 vs. G3 (Correlation = {corr:.2f})",
        "x": 0.5,
        "xanchor": "center",
        "font": {"size": 22, "color": navy}
    },
    plot_bgcolor=background,
    paper_bgcolor=background,
    font=dict(
        family="Arial",
        size=13,
        color=navy
    ),
    margin=dict(l=60, r=40, t=90, b=60)
)
fig.update_xaxes(
    title="G2 (Second Period Grade)",
    dtick=2,
    showgrid=True,
    gridcolor="#E1E5EF"
)
fig.update_yaxes(
    title="G3 (Final Grade)",
    dtick=2,
    showgrid=True,
    gridcolor="#E1E5EF"
)
fig.show()
```

```python
navy = "#202A44"
atlantic_navy = "#13336F"
complementary_blue = "#6C8EBF"
accent_pink = "#D94F70"
background = "#F7F8FC"

summary = df.groupby("failures")["G3"].agg(["mean", "count"]).reset_index()
summary.columns = ["failures", "mean_grade", "student_count"]

summary["failures"] = summary["failures"].astype(str)

fig = px.bar(
    summary,
    x="failures",
    y="mean_grade",
    text="student_count",
    title="Average Final Grade by Number of Previous Failures",
    labels={
        "failures": "Number of Previous Failures",
        "mean_grade": "Average Final Grade (G3)",
        "student_count": "Number of Students"
    },
    color_discrete_sequence=[atlantic_navy],
    template="plotly_white",
    hover_data={"student_count": True, "mean_grade": ":.2f"}
)

fig.update_traces(
    texttemplate="n=%{text}",
    textposition="outside",
    cliponaxis=False,
    marker_line_color=navy,
    marker_line_width=0.5
)

fig.update_layout(
    title={"text": "Average Final Grade by Number of Previous Failures", "x": 0.5, "font": {"size": 22, "color": navy}},
    plot_bgcolor=background,
    paper_bgcolor=background,
    font=dict(family="Arial", size=13, color=navy),
    margin=dict(l=60, r=40, t=90, b=60)
)
fig.update_yaxes(showgrid=True, gridcolor="#E1E5EF", title="Average Final Grade (G3)")
fig.update_xaxes(showgrid=False)
fig.show()
```

```python
df_all = df.copy()
df = df_all[df_all["G3"] > 0].reset_index(drop=True)

print(f"Before: {len(df_all)} students")
print(f"After: {len(df)} students")
```

I excluded 38 students (9.6%) with G3 = 0. Reasons: (1) their attendance was recorded as 0, whereas the average for other students was 6.3, indicating incomplete data rather than perfect attendance; (2) a score of 0 on the final exam likely indicates that they did not take the exam or dropped out of school, rather than a reflection of their learning outcomes; (3) the G2-G3 correlation increased from 0.90 to 0.97 without them.
Limitations: This group has been underperforming since G1 (average 7.5 vs. 11.3) and has a higher failure rate (0.92 vs. 0.27), so excluding them means losing the students at highest risk. The findings apply only to students who took the final exam, and the effect of failures is likely to appear smaller than it actually is.
